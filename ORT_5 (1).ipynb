### 1. Write a program to count the number of notes (R10, R20, R50, R100 or R200) in a given amount.

```python
def count_notes(amount):
    denominations = [200, 100, 50, 20, 10]
    notes = {}

    for denom in denominations:
        if amount >= denom:
            count = amount // denom
            notes[denom] = count
            amount -= count * denom
        else:
            notes[denom] = 0

    return notes, amount  # 'amount' is the leftover (should be 0 if multiple of 10)

# Example
amount = 1000
notes, remainder = count_notes(amount)
print(f"Amount: R{amount}")
for denom, count in notes.items():
    if count > 0:
        print(f"R{denom}: {count}")
if remainder:
    print(f"Remainder: R{remainder} (not divisible by 10)")
```

```python
amount = float(input('How much is the money?:'))
note_denomination = int(input('What note denomination is making up the amount?:'))
if amount < note_denomination:
    print('Amount is less than the denomination specified')
elif note_denomination == 10:
    number_of_notes = amount//10
    print(f'There are {number_of_notes} R10 notes in the amount given') 
elif note_denomination == 20:
    number_of_notes = amount//20
    print(f'There are {number_of_notes} R20 notes in the amount given') 
elif note_denomination == 50:
    number_of_notes = amount//50
    print(f'There are {number_of_notes} R50 notes in the amount given') 
elif note_denomination == 100:
    number_of_notes = amount//100
    print(f'There are {number_of_notes} R100 notes in the amount given') 
elif note_denomination == 200:
    number_of_notes= amount//200
    print(f'There are {number_of_notes} R200 notes in the amount given') 
else:
    print('Invalid denomination')
```

### Write a program to find the sum of odd numbers between a given range

```python
def sum_odd_range(start, end):
    """Return sum of odd integers from start to end (inclusive)."""
    total = 0
    for num in range(start, end + 1):
        if num % 2 != 0:
            total += num
    return total

# User input
start = int(input("Enter start of range: "))
end = int(input("Enter end of range: "))
total = sum_odd_range(start, end)
print(f"Sum of odd numbers from {start} to {end} is: {total}")
```

### Write a program to display prime number between a given range

```python
def is_prime(n):
    """Return True if n is prime, else False."""
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def primes_in_range(start, end):
    """Return list of prime numbers between start and end (inclusive)."""
    return [num for num in range(start, end + 1) if is_prime(num)]

# User input
start = int(input("Enter start of range: "))
end = int(input("Enter end of range: "))
primes = primes_in_range(start, end)
print(f"Prime numbers between {start} and {end}: {primes}") 
```

### A perfect number is a positive integer that is equal to the sum of itd factors. E.g 28 is a perfect number because 1+2+4+7+14 = 28. Write a program using a function to determine whether or not  a given number is a perfect number.

```python
def perfect_number(n):
    number = 0
    for i in range(1, n):
        if n % i == 0:
            number += i
    return number == n
num = int(input("Enter a number: "))
if perfect_number(num):
    print(num, "is a Perfect Number")
else:
    print(num, "is NOT a Perfect Number")
```
